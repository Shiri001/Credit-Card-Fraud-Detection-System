## Random Forest Feature Importance

To complement the statistical analysis, a Random Forest classifier was trained on the training data to estimate feature importance.

Unlike correlation analysis, Random Forest evaluates features based on their contribution to reducing classification error across multiple decision trees.

### Key Observations

- V17, V14, V12, V10, and V16 emerged as the most influential predictors.
- The model's ranking closely aligns with the earlier correlation analysis, validating the importance of these PCA components.

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train[['Time','Amount']] = scaler.fit_transform(
    X_train[['Time', 'Amount']]
)

X_test[['Time','Amount']] = scaler.transform(
    X_test[['Time', 'Amount']]
)

 
````

````python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train,
    y_train
)

print(y_train.value_counts())
print()
print(y_train_smote.value_counts())
````

````python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(
    random_state=42,
    max_iter=1000
)

lr.fit(X_train_smote, y_train_smote)
````

````python
y_pred = lr.predict(X_test)
y_prob = lr.predict_proba(X_test)[:,1]
````

````python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
````

````python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
````

````python
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_prob
)

plt.figure(figsize=(8,6))

plt.plot(thresholds, precision[:-1], label="Precision")
plt.plot(thresholds, recall[:-1], label="Recall")

plt.xlabel("Decision Threshold")
plt.ylabel("Score")
plt.title("Precision and Recall vs Threshold")
plt.legend()

plt.grid(True)

plt.show()
````

````python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = np.arange(0.1, 1.0, 0.1)

for threshold in thresholds:

    y_pred_threshold = (y_prob >= threshold).astype(int)

    precision = precision_score(y_test, y_pred_threshold)
    recall = recall_score(y_test, y_pred_threshold)
    f1 = f1_score(y_test, y_pred_threshold)

    print(f"Threshold: {threshold:.1f}")
    print(f"Precision: {precision:.3f}")
    print(f"Recall: {recall:.3f}")
    print(f"F1 Score: {f1:.3f}")
    print("-"*40)
````

````python
results = []

for threshold in thresholds:

    y_pred_threshold = (y_prob >= threshold).astype(int)

    results.append([
        threshold,
        precision_score(y_test, y_pred_threshold),
        recall_score(y_test, y_pred_threshold),
        f1_score(y_test, y_pred_threshold)
    ])

results = pd.DataFrame(
    results,
    columns=["Threshold", "Precision", "Recall", "F1"]
)

results
````

````python
plt.figure(figsize=(8,6))

plt.plot(results["Threshold"], results["Precision"], marker="o", label="Precision")
plt.plot(results["Threshold"], results["Recall"], marker="o", label="Recall")
plt.plot(results["Threshold"], results["F1"], marker="o", label="F1 Score")

plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Threshold vs Precision, Recall and F1")

plt.grid(True)
plt.legend()

plt.show()
````

# Model Training and Evaluation

## 1. Model Selection

Three classification models were evaluated for credit card fraud detection:

- Logistic Regression
- Random Forest
- XGBoost

The models were trained using the same training and testing methodology to ensure a fair comparison.

Because the dataset is highly imbalanced, **PR-AUC was given more importance than accuracy and ROC-AUC** when comparing models.

---

## 2. Logistic Regression

Logistic Regression was used as the baseline model.

The model achieved:

- ROC-AUC: **0.9699**
- PR-AUC: **0.7249**

Although the model achieved high recall, its precision for the fraud class was very low. This resulted in a large number of false-positive predictions.

Therefore, Logistic Regression was not selected as the final model.

---

## 3. Random Forest

A Random Forest classifier with 100 decision trees was trained on the SMOTE-balanced training data.

Random Forest builds multiple decision trees using different samples and feature subsets, and combines their predictions to produce the final prediction.

The model achieved:

- ROC-AUC: **0.9731**
- PR-AUC: **0.8747**
- Precision: **0.85**
- Recall: **0.84**
- F1-score: **0.84**

Random Forest provided a substantially better precision-recall trade-off than Logistic Regression.

---

## 4. XGBoost

XGBoost was also trained using the same SMOTE-balanced training data.

XGBoost builds trees sequentially, with each new tree attempting to correct the errors of the existing ensemble.

The model achieved:

- ROC-AUC: **0.9802**
- PR-AUC: **0.7984**
- Precision: **0.12**
- Recall: **0.90**
- F1-score: **0.21**

Although XGBoost achieved the highest ROC-AUC and high fraud recall, it produced many false-positive predictions at the default threshold.

Therefore, its precision-recall performance was inferior to Random Forest for this problem.

---

## 5. Model Comparison

| Model | Precision | Recall | F1-score | ROC-AUC | PR-AUC |
|---|---:|---:|---:|---:|---:|
| Logistic Regression | 0.06 | 0.92 | 0.11 | 0.9699 | 0.7249 |
| Random Forest | 0.85 | 0.84 | 0.84 | 0.9731 | **0.8747** |
| XGBoost | 0.12 | 0.90 | 0.21 | **0.9802** | 0.7984 |

### Model Selection

Random Forest was selected as the final model because it achieved the highest **PR-AUC** and provided the strongest overall precision-recall trade-off.

For this highly imbalanced fraud detection problem, PR-AUC is particularly useful because it focuses on the model's ability to identify the minority fraud class while controlling false-positive predictions.

# Threshold Tuning

The default classification threshold of 0.5 was further investigated.

The Random Forest produces a probability of fraud for each transaction:

```text
Fraud probability >= threshold → Fraud
Fraud probability < threshold  → Legitimate
