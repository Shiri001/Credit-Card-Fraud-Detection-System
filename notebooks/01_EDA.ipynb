# Target Variable Analysis

Before exploring the predictor variables, it is important to understand the distribution of the target variable.

The **Class** column indicates whether a transaction is legitimate or fraudulent.

- **0 → Legitimate Transaction**
- **1 → Fraudulent Transaction**

Understanding the class distribution helps identify class imbalance, which is a critical consideration for fraud detection problems.

```python
df["Class"].value_counts()
```

```python
df["Class"].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x = "Class", data = df)

plt.title("Distribution of Legitimate and Fraudulent Transactions")
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    df["Class"].value_counts(),
    labels = ["Legitimate", "Fraud"],
    autopct = "%1.2f%%",
    startangle = 90
)

plt.title("Transaction Class Distribution")

plt.show()
```

### Observations

- The dataset is extremely imbalanced.
- Legitimate transactions account for the vast majority of observations.
- Fraudulent transactions represent only a very small percentage of the dataset.
- This imbalance indicates that special preprocessing and evaluation strategies will be required during model development.

# Numerical Feature Analysis

This section explores the numerical features in the dataset to understand their distributions, identify potential outliers, and assess whether preprocessing techniques such as scaling or transformation may be required.

The analysis begins with the two original features:

- **Time**
- **Amount**

Unlike the PCA-transformed features (V1–V28), these variables retain their original meaning and can provide valuable business insights into transaction behavior.

## Transaction Amount Analysis

The **Amount** feature represents the monetary value of each credit card transaction.

Analyzing this feature helps us understand:

- Typical transaction amounts
- Distribution of transaction values
- Presence of outliers
- Whether scaling will be required before model training

```python
df["Amount"].describe()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["Amount"], bins = 50, kde = True)

plt.title("Distribution of Transaction Amount")

plt.xlabel("Transaction Amount")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,2))

sns.boxplot(x=df["Amount"])

plt.title("Boxplot of Transaction Amount")

plt.show()
```

```python
print("Skewness :", df["Amount"].skew())
print("Kurtosis :", df["Amount"].kurt())
```

### Observations

- Transaction amounts are highly positively skewed, with most transactions involving relatively small monetary values.
- A small number of high-value transactions create a long right tail and appear as statistical outliers.
- These outliers are likely to represent genuine transactions rather than data errors and should not be removed without further investigation.
- Since the `Amount` feature has a much larger numerical scale than the PCA-transformed features, feature scaling will be considered during preprocessing to ensure consistent model training.

## Time Feature Analysis

The `Time` feature represents the number of seconds elapsed between each transaction and the first transaction in the dataset.

Unlike a standard timestamp, it does not indicate the actual date or time of day. Instead, it captures the temporal ordering of transactions, allowing us to study transaction patterns over time.

```python
df["Time"].describe()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(x=df["Time"], bins = 50, kde = True)

plt.title("Distribution of Transaction Time")

plt.xlabel("Time (seconds)")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,2))

sns.boxplot(x=df["Time"])

plt.title("Boxplot of Transaction Time")

plt.show()
```

```python
print("Skewness :", df["Time"].skew())
print("Kurtosis :", df["Time"].kurt())
```

### Observations

- The `Time` feature follows a bimodal distribution rather than a normal distribution, indicating varying transaction activity throughout the recorded period.
- No significant outliers were observed, and the recorded values appear to represent genuine transaction timings.
- Since `Time` has a much larger numerical range than the PCA-transformed features, it will be standardized during preprocessing to ensure consistent feature scaling for machine learning algorithms.


# Time vs Fraud Analysis

The previous section explored the overall distribution of transaction time.

In this section, we investigate whether fraudulent transactions occur uniformly throughout the dataset or exhibit different temporal patterns compared to legitimate transactions.

Understanding these patterns helps determine whether the `Time` feature contains useful predictive information.

```python
df.groupby("Class")["Time"].describe()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Class",
    y="Time",
    data=df
)

plt.xticks([0,1],["Legitimate","Fraud"])

plt.title("Transaction Time by Class")

plt.show()
```

### Observations

- The distributions of transaction time for legitimate and fraudulent transactions overlap considerably.
- Fraudulent transactions occur throughout the recorded time period rather than being concentrated in a single interval.
- The median transaction time for fraudulent transactions is slightly lower than that of legitimate transactions, but the difference is not substantial.
- Since both classes exhibit similar distributions, the `Time` feature alone is unlikely to distinguish fraudulent transactions effectively. However, it may still contribute useful information when combined with other features.

```python
plt.figure(figsize=(8,5))

sns.violinplot(
    x="Class",
    y="Time",
    data=df
)

plt.xticks([0,1],["Legitimate","Fraud"])

plt.title("Distribution of Transaction Time by Class")

plt.show()
```

### Observations

- Both legitimate and fraudulent transactions exhibit bimodal distributions, indicating two periods of higher transaction activity.
- Fraudulent transactions appear slightly more concentrated during the earlier portion of the recorded timeline.
- However, the distributions overlap substantially across the entire time range.
- The `Time` feature alone does not provide sufficient discriminatory power for identifying fraudulent transactions, though it may contribute useful information when combined with other features.

```python
#Now lets see the correlation between the feature formed using PCA and our class

corr = df.corr(numeric_only = True)
corr["Class"].sort_values()
```

```python
corr["Class"].sort_values(key = abs, ascending = False)
```

```python
top_features = ["V17", "V14", "V12", "V10", "V11", "V4"]

plt.figure(figsize=(18, 10))

for i, feature in enumerate(top_features):
    plt.subplot(2, 3, i + 1)

    sns.boxplot(
        x="Class",
        y=feature,
        data=df
    )

    plt.xticks([0, 1], ["Legitimate", "Fraud"])
    plt.title(feature)

plt.tight_layout()
plt.show()
```

## Distribution of Highly Correlated PCA Features

To better understand the statistical relationships identified through correlation analysis, the distributions of the most strongly correlated PCA components were compared across legitimate and fraudulent transactions using boxplots.

### Key Observations

- V17, V14, V12, and V10 exhibit substantially lower values for fraudulent transactions, consistent with their negative correlations.
- V11 and V4 display noticeably higher values for fraudulent transactions, confirming their positive correlations.
- Compared with the original `Time` and `Amount` features, these PCA components show much clearer separation between legitimate and fraudulent transactions.
- Although some overlap remains, these features demonstrate considerably stronger discriminatory power and are expected to contribute significantly during model training.

# Relationship Among Important Features

After identifying the principal components most strongly associated with fraudulent transactions, the next step is to examine the relationships among these features.

This analysis helps determine whether the important variables provide complementary information or whether some of them are highly correlated with each other.

Understanding these relationships is particularly useful for linear models such as Logistic Regression, where strong multicollinearity may affect model stability.

```python
important = ["V17", "V14", "V12", "V10", "V11", "V4", "Class"]

plt.figure(figsize=(8,6))

sns.heatmap(
    df[important].corr(),
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Correlation Among Important Features")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.kdeplot(
    data = df[df["Class"] == 0],
    x = "V17",
    label="Legitimate",
    fill = True,
    alpha = 0.4
)

sns.kdeplot(
    data = df[df["Class"] == 1],
    x = "V17",
    label="Fraud",
    fill = True,
    alpha = 0.4
)

plt.title("Distribution of V17 by Transaction Class")
plt.legend()

plt.show()
```

## Density Distribution of Important PCA Features

Kernel Density Estimation (KDE) plots were used to compare the full distributions of the most informative PCA components across legitimate and fraudulent transactions.

Unlike boxplots, KDE plots reveal the overall shape, concentration, and overlap of the distributions, providing deeper insight into each feature's ability to distinguish fraudulent behavior.

The analysis focuses on identifying whether fraudulent transactions exhibit consistent shifts in the feature distributions relative to legitimate transactions.

## Distribution Analysis of V17

The KDE plot reveals a clear shift between legitimate and fraudulent transactions.

### Key Observations

- Legitimate transactions are highly concentrated around V17 values close to zero.
- Fraudulent transactions exhibit substantially lower V17 values and a much wider distribution.
- The overlap between the two classes is relatively small compared with the original `Time` and `Amount` features.
- The clear leftward shift of the fraud distribution visually confirms the negative correlation observed during the correlation analysis.
- Although V17 alone cannot perfectly separate the classes, it provides strong discriminatory information and is expected to be one of the most influential predictors during model training.
